```python
from pathlib import Path
import zipfile 

PROJECT_ROOT = Path.cwd().parents[1]

DATA_ROOT = PROJECT_ROOT / "tennis_data"

print("Project roo: ", PROJECT_ROOT)
print("Data root: ", DATA_ROOT)
```

```python
EXTRACT_ROOT = DATA_ROOT / "extracted"

EXTRACT_ROOT.mkdir(parents=True, exist_ok=True)
```

```python
#test

zip_files = sorted(DATA_ROOT.glob("*.zip"))

print("number of zip files: ", len(zip_files))
print(zip_files[:3])
```

```python
zip_files = sorted(DATA_ROOT.glob("*.zip"))

for zip_path in zip_files:
    day_name = zip_path.stem
    day_folder = EXTRACT_ROOT / day_name
    day_folder.mkdir(parents=True, exist_ok=True)

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:

        for file in zip_ref.namelist():
            if file.endswith(".parquet"):
                parts = Path(file).parts
                raw_index =parts.index("raw")
                relative_path = Path(*parts[raw_index + 1:])

                output_file = day_folder / relative_path
                output_file.parent.mkdir(parents=True, exist_ok=True)

                with zip_ref.open(file) as source:
                    with open(output_file, "wb") as target:
                        target.write(source.read())
```
