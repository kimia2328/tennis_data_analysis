```python


home_team=pl.read_parquet(
    PROJECT_PATH/"tennis_data"/"Data"/"away_and_home_team"/"home_team.parquet" 
    )
away_team=pl.read_parquet(
    PROJECT_PATH/"tennis_data"/"Data"/"away_and_home_team"/"away_team.parquet"
)
```

```python
home_team.select("country").head()
```

```python
away_team.select("country").head()
```

```python
countries = pl.concat(
    [
        home_team.select("country"),
        away_team.select("country")
    ]
)
```

```python
countries_clean = countries.filter(
    pl.col("country").is_not_null()
)
```

```python
distinct_countries = countries_clean.select(
    pl.col("country").n_unique()
)

distinct_countries
```

## Objective

The objective of this analysis is to determine the number of unique countries represented by tennis players in the dataset.

## Data Used

The player information was extracted from the following tables:

- MatchHomeTeamInfo
- MatchAwayTeamInfo

The relevant feature used for this analysis was:

- country

This column represents the player's country/nationality.

## Data Preparation

Since a player can appear as either a home player or an away player, both player datasets were combined to include all participants.

Missing country values were removed because they do not provide valid country information.

## Important Note

The dataset also contains a country column in the MatchVenueInfo table.

However, this column represents the location of the venue where the match was played, not the player's country.

Therefore, venue country information was not used in this analysis because the goal was to identify the countries represented by tennis players.

## Method

The number of distinct countries was calculated by counting the unique values in the player's country column.

## Result

The analysis shows that the dataset contains:

101 distinct countries

represented by tennis players.

## Conclusion

The dataset includes players from a diverse range of countries, with 101 unique player countries represented.

This result reflects the nationality distribution of players included in the dataset, rather than the countries where tournaments were hosted.
