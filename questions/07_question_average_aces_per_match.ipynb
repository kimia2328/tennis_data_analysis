# Question 7: What is the average number of aces per match?

Using `statistics`

```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
MINI_PROJECT_ROOT = Path.cwd().parents[2]

DATA_ROOT = (
    MINI_PROJECT_ROOT
    / "Tennis Schema"
    / "tennis_data"
)

PROCESSED_ROOT = DATA_ROOT / "processed"

TEAM_DATA_ROOT = DATA_ROOT / "Data"
```

```python
statistics = pd.read_parquet(PROCESSED_ROOT / "statistics_clean.parquet")

statistics.loc[statistics["statistic_name"].str.contains("ace", case=False, na=False), "statistic_name"].value_counts()
```

```python
# Keep Match-Level Ace Totals

aces = statistics[(statistics["statistic_name"] == "aces")& (statistics["period"] == "ALL")]

aces[["match_id", "home_value", "away_value"]].head()
```

```python
# Keep the Latest Ace Snapshot per Match

aces_latest = (aces.sort_values("snapshot_date").drop_duplicates("match_id", keep="last"))

aces_latest[["match_id", "home_value", "away_value"]].head()
```

```python
# Total aces per match

aces_latest["total_aces"] = (
    aces_latest["home_value"] + aces_latest["away_value"]
)

aces_latest[["match_id", "home_value", "away_value", "total_aces"]].head()
```

```python
# Get the average

average_aces_per_match = aces_latest["total_aces"].mean().round(2)

print(f"On average, a tennis match has {average_aces_per_match} aces")
```
