# Question 17: What is the average number of breaks of serve per match?

Using `tennis_power`

```python
import pandas as pd
from pathlib import Path
```

```python
MINI_PROJECT_ROOT = Path.cwd().parents[2]

DATA_ROOT = (
    MINI_PROJECT_ROOT
    / "Tennis Schema"
    / "tennis_data"
)

PROCESSED_ROOT = DATA_ROOT / "processed"
```

```python
power = pd.read_parquet(PROCESSED_ROOT / "tennis_power_clean.parquet")

power[["match_id", "break_occurred"]].head()
```

```python
power.columns.tolist()
```

```python
# Keep the Last Snapthot

power_latest_game = (
    power
    .sort_values("snapshot_date")
    .drop_duplicates(
        subset=["match_id", "set_num", "game_num"],
        keep="last"
    )
)

power_latest_game["break_occurred"].value_counts()
```

```python
# Calculate the Average Number of Breaks per Match

breaks_per_match = (power_latest_game.groupby("match_id")["break_occurred"].sum())

average_breaks_per_match = breaks_per_match.mean()

print(f"The average number of breaks of serve per match is {average_breaks_per_match:.2f}")
```
