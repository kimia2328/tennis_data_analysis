```python
from pathlib import Path

data_path = Path("../tennis_data")

zip_files = list(data_path.glob("*.zip"))

print("number of zip files:",len(zip_files))
```

```python
import zipfile

with zipfile.ZipFile(zip_files[0], "r") as zip_file:
    files = zip_file.namelist()

print(files[:10])
```

```python
extract_path = data_path / "extracted"

extract_path.mkdir(exist_ok=True)

print("Extraction path:",extract_path)
```

```python
for zip_path in zip_files:
    print(zip_path.name)
```

```python
for zip_path in zip_files:
    
    day_name = zip_path.stem
    day_path = extract_path / day_name
    day_path.mkdir(exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as zip_file:

        for file_name in zip_file.namelist():

            if file_name.endswith(".parquet"):

                output_file = day_path / Path(file_name).name

                with zip_file.open(file_name) as source:
                    with open(output_file, "wb") as target:
                        target.write(source.read())

print("Extraction completed.")
```

```python
days = [path for path in extract_path.iterdir() if path.is_dir()]

print("Number of extracted days:", len(days))
```
